```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

data = [
    ('2016-10-03', 772.559998),
    ('2016-10-04', 776.429993),
    ('2016-10-05', 776.469971),
    ('2016-10-06', 776.859985),
    ('2016-10-07', 775.080017)
]

# builds dataframe with pandas
df = pd.DataFrame(data, columns=['Date', 'Close'])
df['Date'] = pd.to_datetime(df['Date'])
df.set_index('Date', inplace=True)

# plots points
fig, ax = plt.subplots()
ax.plot(df.index, df['Close'], color='red', marker='o', label='Close')

# sets ticks and formatting
ax.set_xticks(df.index)
ax.xaxis.set_major_formatter(mdates.DateFormatter('%Y-%m-%d'))

# sets titles and labels
ax.set_title('Closing stock value of Alphabet inc.')
ax.set_xlabel('Date')
ax.set_ylabel('Closing Value')

# configures grid
ax.grid(which='major', linewidth=1, color='red')
ax.minorticks_on()
ax.grid(which='minor', linewidth=0.6, linestyle=':')

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# data import
languages = ['Java', 'Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8.0, 7.7, 6.7]

#plots data
fig, ax = plt.subplots()
bars = ax.bar(languages, popularity, color='blue')

#titles and labels
ax.set_title('Popularity of Programming Languages Worldwide, Oct 2017 compared to a year ago')
ax.set_xlabel('Languages')
ax.set_ylabel('Popularity')

# major ticks
ax.set_xticks(range(len(languages)))
ax.set_xticklabels(languages)
ax.set_yticks(range(0, 21, 5))  # doesnt show values past 20, like the example

# minor ticks
ax.set_yticks(range(0, 21, 1), minor=True)
ax.set_xticks([i + 0.5 for i in range(len(languages) - 1)], minor=True)

# sets ticks
ax.minorticks_on()

# onfigures grid
ax.grid(which='major', axis='both', linewidth=1, color='red')
ax.grid(which='minor', axis='both', linestyle=':', linewidth=0.5)

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

#data import
languages = ['Java', 'Python', 'PHP', 'JS', 'C#', 'C++']
popularity = [22.2, 17.6, 8.8, 8.0, 7.7, 6.7]

# Plotting
fig, ax = plt.subplots()
ax.barh(languages, popularity, color='green')

# titles and labels
ax.set_title('Popularity of Programming Languages Worldwide, Oct 2017 compared to a year ago')
ax.set_xlabel('Popularity')
ax.set_ylabel('Language')

# labels x axis up to 20
ax.set_xlim(0, 25)
ax.set_xticks([0, 5, 10, 15, 20])
ax.set_xticks(range(0, 26, 1), minor=True)
plt.minorticks_on()

plt.grid(which='major', linestyle='-', linewidth=0.8, color='red')
plt.grid(which='minor', linestyle=':', linewidth=0.5)

plt.tight_layout()
plt.show()
```
